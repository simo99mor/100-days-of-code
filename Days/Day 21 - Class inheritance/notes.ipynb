```python
class Animal:
    def __init__(self):
        self.num_eyes = 2
    
    def breathe(self):
        print("Inhale, exhale.")
```

```python
class Fish(Animal):   # To inherit from the class Animals, I have to put it into the parenthesis ()
    def __init__(self):
        super().__init__()  #  the call to super() in the initialiser is recommended, but not strictly required
        # In this way I'm inheriting all of the attributes and methods from the super class (Animal)

    # I want to etend the functionality of the method "breath" of the super class
    def breathe(self):
        super().breathe()  # I take what's already present in the method
        print("doing this underwater.")  # I add this line

    def swim(self):
        print("moving in water.")
```

```python
nemo = Fish()
```

```python
nemo.swim()

nemo.breathe()
print(nemo.num_eyes)
```
